```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='Discount', y='Profit', hue='Category', size='Sales', sizes=(20, 200))
plt.title('Discount vs. Profit by Category and Sales Size')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid()
plt.show()
```

From the correlation matrices, let's look for a pair of variables where the Pearson and Spearman correlation coefficients differ significantly. A notable difference suggests a non-linear relationship or the presence of outliers. Let's inspect 'Sales' and 'Profit' as an example, as the previous scatter plot hinted at a complex relationship.

# Task 4

```python
numerical_df = df.select_dtypes(include=[np.number])

pearson_corr = numerical_df.corr(method='pearson')
spearman_corr = numerical_df.corr(method='spearman')

plt.figure(figsize=(12, 8))
sns.scatterplot(data=numerical_df, x='Sales', y='Profit')
plt.title('Sales vs. Profit by Category and Discount Size')
plt.grid()
plt.show()

print("Pearson Correlation Matrix:")
print(pearson_corr)

print("\nSpearman Correlation Matrix:")
print(spearman_corr)


```

```python
sns.pairplot(numerical_df, diag_kind='kde')
plt.title('Pairwise Scatter Plots of Numerical Variables')
plt.show()
```

The Pearson correlation for 'Sales' and 'Profit' is significantly lower than the Spearman correlation. This noticeable disagreement indicates that while there is a monotonic relationship (as one tends to increase, the other tends to increase or stay consistent, as captured by Spearman's rank correlation), the relationship is not strictly linear or is heavily influenced by outliers. The scatter plot confirms this, showing a strong positive trend for lower sales values, but with several high sales values having relatively low or even negative profits (outliers), which pulls the Pearson coefficient down. Pearson is sensitive to the magnitude of changes and outliers, while Spearman focuses on the rank order.

# Task 6

```python
df.index = pd.to_datetime(df.index)
monthly = df['Sales'].resample('ME').sum()
monthly_totals_rolling = monthly.rolling(window=3).mean()
monthly.plot(label='Monthly sales')
monthly_totals_rolling.plot(label='month rolling avg')
plt.legend()
```

The rolling average smooths out short-term fluctuations present in the raw monthly sales data, making it easier to identify underlying trends and long-term patterns. This smoothing helps to reduce noise and highlight whether sales are generally increasing, decreasing, or remaining stable over time.

```python
df['Is_Profitable'] = df['Profit'] > 0

g = sns.relplot(
    data=df,
    x='Discount',
    y='Profit',
    col='Region',
    hue='Is_Profitable',
    kind='scatter',
    col_wrap=2,
    height=4,
    aspect=1.2,
    palette={True: 'green', False: 'red'},
    facet_kws={'sharex': True, 'sharey': True}
)
g.set_axis_labels("Discount", "Profit")
g.set_titles("Region: {col_name}")
g.fig.suptitle("Discount vs. Profit by Region and Profitability", y=1.02, fontsize=16)
plt.tight_layout()
plt.show()
```

Observing the plots, the **Central** region appears to have a visibly different discount-profit relationship compared to others. While all regions show a general trend of decreasing profit (or increasing losses) with higher discounts, the Central region exhibits a particularly strong negative impact of discounts on profit, with a significant number of heavily discounted items resulting in substantial losses. The distribution of profitable transactions at higher discount rates also seems narrower compared to other regions, indicating less success in offsetting discounts with high sales volumes to maintain profitability.

```python
# Raw scatter plot showing heavy overplotting
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Discount', y='Profit')
plt.title('Raw Discount vs. Profit Scatter (Heavy Overplotting)')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid()
plt.show()
```

For approximately 10,000 data points, a `hexbin` plot is a better choice in a scatter plot. `hexbin` effectively visualizes the density of points in different regions, which is crucial when individual points are no longer distinguishable due to overplotting. It helps identify areas of high concentration of data.

```python
plt.figure(figsize=(12, 8))
sns.scatterplot(data=df, x='Discount', y='Profit',alpha = 0.9)

most_profitable = df.loc[df['Profit'].idxmax()]

largest_loss = df.loc[df['Profit'].idxmin()]

plt.scatter(most_profitable['Discount'], most_profitable['Profit'], color='red', label='Most Profitable')
plt.scatter(largest_loss['Discount'], largest_loss['Profit'], color='blue', label='Largest Loss')

plt.legend()

plt.title('Discount vs. Profit (Hexbin)')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.grid()
plt.show()
```
